## Question 5: Data Filtering Using Conditions

### 5.1 Extract all patients older than 50 years

```python
older_50 = df[df['age'] > 50]
older_50
```

### 5.2 Extract patients who have heart disease and a cholesterol level greater than 240

```python
hd_high_chol = df[(df['target'] == 1) & (df['chol'] > 240)]
hd_high_chol
```

### 5.3 Count how many patients satisfy both conditions

```python
print('Number of patients with heart disease and cholesterol > 240:', len(hd_high_chol))
```
